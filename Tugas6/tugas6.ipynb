```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Parameter distribusi Poisson
lambda_demand = 15  # rata-rata permintaan per hari
n_simulations = 100  # jumlah hari yang disimulasikan

# Simulasi permintaan harian
np.random.seed(42)  # untuk reproduktifitas
daily_demand_poisson = np.random.poisson(lam=lambda_demand, size=n_simulations)

# Visualisasi: Histogram
plt.figure(figsize=(10, 6))
plt.hist(daily_demand_poisson, bins=range(min(daily_demand_poisson), max(daily_demand_poisson) + 2), 
         density=True, alpha=0.7, color='skyblue', label='Simulasi Poisson')
# Plot distribusi teoritis
x = np.arange(0, max(daily_demand_poisson) + 1)
plt.plot(x, poisson.pmf(x, lambda_demand), 'ro-', label='Poisson Teori')
plt.title('Distribusi Permintaan Barang Harian (Poisson)')
plt.xlabel('Jumlah Permintaan (unit)')
plt.ylabel('Probabilitas')
plt.legend()
plt.grid(True)
plt.show()

# Visualisasi: Tren permintaan harian
plt.figure(figsize=(10, 6))
plt.plot(range(1, n_simulations + 1), daily_demand_poisson, 'b-', label='Permintaan Harian')
plt.title('Tren Permintaan Barang Harian (Poisson)')
plt.xlabel('Hari ke-n')
plt.ylabel('Jumlah Permintaan (unit)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
simulated_mean_poisson = np.mean(daily_demand_poisson)
simulated_variance_poisson = np.var(daily_demand_poisson)
theoretical_mean_poisson = lambda_demand
theoretical_variance_poisson = lambda_demand
print(f"Poisson - Rata-rata simulasi: {simulated_mean_poisson:.2f} unit")
print(f"Poisson - Rata-rata teoritis: {theoretical_mean_poisson:.2f} unit")
print(f"Poisson - Variansi simulasi: {simulated_variance_poisson:.2f}")
print(f"Poisson - Variansi teoritis: {theoretical_variance_poisson:.2f}")
```

```python
from scipy.stats import randint

# Parameter distribusi Uniform diskrit
a = 10  # batas bawah
b = 20  # batas atas
n_simulations = 100  # jumlah hari yang disimulasikan

# Simulasi permintaan harian
np.random.seed(42)  # untuk reproduktifitas
daily_demand_uniform = np.random.randint(low=a, high=b + 1, size=n_simulations)

# Visualisasi: Histogram
plt.figure(figsize=(10, 6))
plt.hist(daily_demand_uniform, bins=range(a, b + 2), density=True, alpha=0.7, color='lightgreen', 
         label='Simulasi Uniform')
# Plot distribusi teoritis
x = np.arange(a, b + 1)
plt.plot(x, randint.pmf(x, a, b + 1), 'go-', label='Uniform Teori')
plt.title('Distribusi Permintaan Barang Harian (Uniform Diskrit)')
plt.xlabel('Jumlah Permintaan (unit)')
plt.ylabel('Probabilitas')
plt.legend()
plt.grid(True)
plt.show()

# Visualisasi: Tren permintaan harian
plt.figure(figsize=(10, 6))
plt.plot(range(1, n_simulations + 1), daily_demand_uniform, 'g-', label='Permintaan Harian')
plt.title('Tren Permintaan Barang Harian (Uniform Diskrit)')
plt.xlabel('Hari ke-n')
plt.ylabel('Jumlah Permintaan (unit)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
simulated_mean_uniform = np.mean(daily_demand_uniform)
simulated_variance_uniform = np.var(daily_demand_uniform)
theoretical_mean_uniform = (a + b) / 2
theoretical_variance_uniform = ((b - a + 1) ** 2 - 1) / 12
print(f"Uniform - Rata-rata simulasi: {simulated_mean_uniform:.2f} unit")
print(f"Uniform - Rata-rata teoritis: {theoretical_mean_uniform:.2f} unit")
print(f"Uniform - Variansi simulasi: {simulated_variance_uniform:.2f}")
print(f"Uniform - Variansi teoritis: {theoretical_variance_uniform:.2f}")
```
