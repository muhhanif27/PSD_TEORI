## Original Code (CES)

```python
import numpy as np
import matplotlib.pyplot as plt
# Parameters
initial_temperature = 100    # Initial object temperature
ambient_temperature = 25     # Room temperature
cooling_rate = 0.05          # Rate of heat loss
time_steps = 200             # Total simulation time
dt = 0.1                     # Small time step

# Initialize arrays
time = np.arange(0, time_steps * dt, dt)
temperature = np.zeros(len(time))
temperature[0] = initial_temperature

# Euler's Method to Solve dT/dt = -k (T - T_ambient)
for t in range(1, len(time)):
    dT = -cooling_rate * (temperature[t-1] - ambient_temperature) * dt
    temperature[t] = temperature[t-1] + dT

# Plot results
plt.plot(time, temperature, label="Object Temperature")
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Newton's Law of Cooling Simulation")
plt.legend()
plt.grid()
plt.show()
```

## Change the cooling rate and observe the effect on temperature decay

```python
# Parameters
initial_temperature = 100    # Initial object temperature
ambient_temperature = 25     # Room temperature
cooling_rates = [0.025, 0.05, 0.1]  # Different cooling rates to compare
time_steps = 400             # Simulation time
dt = 0.1                     # Small time step

# Define colors for each cooling rate
colors = ['blue', 'green', 'red']  # Different colors for each curve

# Initialize time array
time = np.arange(0, time_steps * dt, dt)

# Simulate for each cooling rate
plt.figure(figsize=(10, 6))
for i, cooling_rate in enumerate(cooling_rates):
    # Initialize temperature array
    temperature = np.zeros(len(time))
    temperature[0] = initial_temperature

    # Euler's Method to Solve dT/dt = -k (T - T_ambient)
    for t in range(1, len(time)):
        dT = -cooling_rate * (temperature[t-1] - ambient_temperature) * dt
        temperature[t] = temperature[t-1] + dT

    # Plot results for this cooling rate with a specific color
    plt.plot(time, temperature, label=f"Cooling Rate = {cooling_rate}", 
             color=colors[i], linewidth=2)

# Add plot details
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Newton's Law of Cooling Simulation (Different Cooling Rates)")
plt.legend()
plt.grid()
plt.show()
```

## Simulate heating instead of cooling (modify differential equation).


```python
# Parameters
initial_temperature = 100    # Suhu awal benda
ambient_temperature = 25     # Suhu lingkungan (hanya untuk referensi di grafik)
target_temperature = 150     # Suhu target untuk pemanasan
heating_rate = 0.05          # Laju pemanasan
time_steps = 200             # Total waktu simulasi
dt = 0.1                     # Langkah waktu kecil

# Inisialisasi array
time = np.arange(0, time_steps * dt, dt)
temperature = np.zeros(len(time))
temperature[0] = initial_temperature

# Metode Euler untuk menyelesaikan dT/dt = k (T_target - T) untuk pemanasan
for t in range(1, len(time)):
    dT = heating_rate * (target_temperature - temperature[t-1]) * dt
    temperature[t] = temperature[t-1] + dT

# Plot hasil
plt.plot(time, temperature, label="Suhu Benda", color='red')
plt.axhline(y=target_temperature, color='orange', linestyle='--', label=f'Suhu Target ({target_temperature}°C)')
plt.axhline(y=ambient_temperature, color='blue', linestyle='--', label=f'Suhu Lingkungan ({ambient_temperature}°C)')
plt.xlabel("Waktu")
plt.ylabel("Suhu (°C)")
plt.title("Simulasi Hukum Pemanasan Newton")
plt.legend()
plt.grid()
plt.show()
```

## Use a real-world dataset from Kaggle (e.g., weather data) to simulate continuous temperature changes.


```python
import pandas as pd

# Baca dataset dari file CSV
df = pd.read_csv('london_weather.csv')

# Ubah kolom 'date' menjadi format datetime
df['date'] = pd.to_datetime(df['date'], format='%Y%m%d')

# Filter data hanya untuk tahun 2020
df_2020 = df[df['date'].dt.year == 2020].copy()

# Pastikan tidak ada data di luar 2020
df_2020 = df_2020[df_2020['date'] <= '2020-12-31']

# Ambil data waktu dan suhu lingkungan (mean_temp)
time = df_2020['date']
ambient_temperatures = df_2020['mean_temp'].fillna(method='ffill').values  # Isi NaN dengan nilai sebelumnya

# Parameter simulasi
initial_temperature = 100  # Suhu awal benda
cooling_rate = 0.05        # Laju pendinginan
dt = 1.0                   # Langkah waktu (1 hari, karena data harian)
time_steps = len(ambient_temperatures)  # Jumlah langkah sesuai panjang data

# Inisialisasi array suhu benda
temperature = np.zeros(time_steps)
temperature[0] = initial_temperature

# Metode Euler untuk simulasi suhu benda dengan suhu lingkungan yang berubah
for t in range(1, time_steps):
    dT = -cooling_rate * (temperature[t-1] - ambient_temperatures[t-1]) * dt
    temperature[t] = temperature[t-1] + dT

# Plot hasil
plt.figure(figsize=(12, 6))
plt.plot(time, temperature, label="Suhu Benda", color='red')
plt.plot(time, ambient_temperatures, label="Suhu Lingkungan (Data Cuaca 2020)", color='blue', linestyle='--')
plt.xlabel("Tanggal")
plt.ylabel("Suhu (°C)")
plt.title("Simulasi Perubahan Suhu Benda dengan Data Cuaca Tahun 2020")
plt.legend()
plt.grid()
plt.xticks(rotation=45)
plt.xlim(pd.Timestamp('2020-01-01'), pd.Timestamp('2020-12-31'))  # Batasi sumbu x hanya untuk 2020
plt.tight_layout()

# Tampilkan plot
plt.show()
```

## Compare CES vs. DES for similar systems.


```python
#CES
# Parameters
initial_temperature = 100    # Initial object temperature
ambient_temperature = 25     # Room temperature
cooling_rate = 0.05          # Rate of heat loss
time_steps = 200             # Total simulation time
dt = 0.1                     # Small time step

# Initialize arrays
time = np.arange(0, time_steps * dt, dt)
temperature = np.zeros(len(time))
temperature[0] = initial_temperature

# Euler's Method to Solve dT/dt = -k (T - T_ambient)
for t in range(1, len(time)):
    dT = -cooling_rate * (temperature[t-1] - ambient_temperature) * dt
    temperature[t] = temperature[t-1] + dT

# Plot results
plt.plot(time, temperature, label="Object Temperature")
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("CES: Newton's Law of Cooling Simulation")
plt.legend()
plt.grid()
plt.show()
```

```python
# DES
# Parameters
initial_temperature = 100    # Initial object temperature
ambient_temperature = 25     # Room temperature
cooling_rate = 0.05          # Rate of heat loss
time_steps = 200             # Total simulation time
pause_duration = 2           # Duration to pause cooling after an event (in time units)

# Initialize lists to store time and temperature at events
time_des = [0]  # Start at t=0
temperature_des = [initial_temperature]  # Start at 100°C
current_temp = initial_temperature
current_time = 0

# Define temperature thresholds for events (every 10°C drop)
temp_thresholds = np.arange(90, ambient_temperature - 5, -10)  # 90, 80, ..., 30

# Simulate discrete events
while current_time < time_steps:
    # Find the next temperature threshold to trigger an event
    for target_temp in temp_thresholds:
        if current_temp > target_temp >= ambient_temperature:
            # Calculate time to reach the target temperature using analytical solution
            t = -(1 / cooling_rate) * np.log((target_temp - ambient_temperature) / (current_temp - ambient_temperature))
            next_event_time = current_time + t

            # If the event time is within the simulation duration, add it
            if next_event_time <= time_steps:
                time_des.append(next_event_time)
                temperature_des.append(target_temp)
                current_time = next_event_time
                current_temp = target_temp

                # Event: Pause cooling for 'pause_duration' time units (temperature remains constant)
                if current_time + pause_duration <= time_steps:
                    time_des.append(current_time + pause_duration)
                    temperature_des.append(current_temp)  # Temperature stays the same during pause
                    current_time += pause_duration
                else:
                    break
            else:
                break
        else:
            continue
    break  # Exit loop after processing all thresholds

# Add final point at the end of simulation
if current_time < time_steps:
    # Calculate final temperature at the end of simulation
    final_temp = ambient_temperature + (current_temp - ambient_temperature) * np.exp(-cooling_rate * (time_steps - current_time))
    time_des.append(time_steps)
    temperature_des.append(final_temp)

# Convert to numpy arrays for plotting
time_des = np.array(time_des)
temperature_des = np.array(temperature_des)

# Plot results
plt.figure(figsize=(10, 6))
plt.plot(time_des, temperature_des, 'o-', label="Object Temperature (DES)", color='green', markersize=8)
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Discrete Event Simulation (DES): Heat Transfer with Cooling Pauses")
plt.legend()
plt.grid()
plt.show()

# Print number of events
events = len([t for i, t in enumerate(temperature_des) if i % 2 == 0]) - 1  # Count event triggers (not pauses)
print(f"Jumlah event (penurunan suhu 10°C): {events}")
```
